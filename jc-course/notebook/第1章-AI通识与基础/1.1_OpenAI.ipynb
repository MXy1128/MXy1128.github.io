d# 2.访问模型

```python
print("正在调用大模型。。。")
response = client.chat.completions.create(
    model = "deepseek-chat",
    messages=[
        {"role":"system","content":"你是一名友好的AI助教。"},
        {"role":"user", "content":"你好，你是谁？"}
    ],
    stream = False,
    temperature=0.9,
    )
print("成功调用大模型。。。")
```

# 3.获取模型响应结果
